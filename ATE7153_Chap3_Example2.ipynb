### Example on Regression

```python
import pandas as pd
from sklearn.model_selection import train_test_split as split
from sklearn.linear_model import LinearRegression
from sklearn.neighbors import KNeighborsRegressor
```

```python
df = pd.read_csv("https://raw.githubusercontent.com/wooihaw/datasets/main/synchronous_machine.csv")
df.head()
```

```python
df.info()
```

```python
df.isna().sum()
```

```python
X = df.drop(columns=['If'])
y = df['If']

X_train, X_test, y_train, y_test = split(X, y, test_size=0.25, random_state=42)
```

```python
lnr = LinearRegression().fit(X_train, y_train)
print(f'lnr R2 score: {lnr.score(X_test, y_test):.3f}')
```

```python
knn = KNeighborsRegressor().fit(X_train, y_train)
print(f'knn R2 score: {knn.score(X_test, y_test):.3f}')
```
